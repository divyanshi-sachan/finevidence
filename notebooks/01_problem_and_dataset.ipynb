## Project objective and motivation

Financial-report questions can require evidence from both narrative text and tables, correct interpretation of reporting periods and units, and one or more calculations. FinEvidence investigates an answer format that includes a numerical response, the supporting evidence, and an interpretable calculation trail when applicable.

The project is a research prototype for evidence inspection. It is not financial advice, an investment-recommendation system, or a guarantee of correctness for arbitrary financial documents.

## Problem definition and research questions

The research problem is how an NLP-based system can identify relevant report evidence, perform transparent numerical reasoning, and present a verifiable financial answer while signalling insufficient or ambiguous evidence.

The M1 research questions are:

1. How effectively can relevant financial evidence be identified from tables and narrative text?
2. How accurately can financial questions requiring multi-step calculation be answered?
3. Does explicit evidence retrieval and deterministic calculation improve reliability over a simpler baseline?
4. Which failures occur around values, units, reporting periods, evidence selection, and arithmetic?
5. Can the system recognize questions that cannot be answered reliably from the available evidence?

## Target users and use cases

Primary users are financial analysts and financial-research professionals. They need to compare financial performance across periods, retrieve the report evidence behind a numerical answer, inspect the calculation, and recognize when an answer is not sufficiently grounded. Corporate-finance teams, business analysts, and students are secondary users for the prototype.

The system will assist evidence inspection; it will not make automated investment or financial decisions.
